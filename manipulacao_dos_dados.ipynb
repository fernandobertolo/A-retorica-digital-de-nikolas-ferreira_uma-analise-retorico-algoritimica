```python
# Anota as palavras com as tags morfológicas

import spacy
import os

# 1. Carregamento do modelo de linguagem
nlp = spacy.load("pt_core_news_lg")

def anotar_corpus_para_antconc(texto_entrada):
    """
    Processa um texto e retorna a versão anotada morfologicamente no formato palavra_TAG.
    """
    # Aumenta o limite de caracteres se os textos forem muito longos
    nlp.max_length = 2000000

    doc = nlp(texto_entrada)
    tokens_anotados = []

    for token in doc:
        # Exclui espaços em branco isolados para não quebrar a formatação
        if not token.is_space:
            # Concatena a palavra e a classe gramatical (POS tag) com um sublinhado
            # token.text: palavra original
            # token.pos_: etiqueta morfológica universal (ex: NOUN, VERB, ADJ)
            token_formatado = f"{token.text}_{token.pos_}"
            tokens_anotados.append(token_formatado)

    # Reconstrói o texto com espaços separando os tokens anotados
    return " ".join(tokens_anotados)

transcricoes_tageadas = []
for i, transcricao in enumerate(transcricoes):
    print(f'Tageando texto {i} de {len(transcricoes)}')
    transcricao_tag = anotar_corpus_para_antconc(transcricao)
    transcricoes_tageadas.append(transcricao_tag)
```

```python
for i, transc_tag in enumerate(transcricoes_tageadas):
    caminho_destino = os.path.join("corpus_antconc", f"anotado_{i:03d}.txt")

    # Escrita com codificação estrita UTF-8
    with open(caminho_destino, 'w', encoding='utf-8') as f:
        f.write(transc_tag)

print(f"Processamento concluído. Ficheiros salvos em (vc sabe onde ;P).")

# Execução do lote
# processar_diretorio("pasta_textos_brutos", "pasta_textos_anotados")
```
